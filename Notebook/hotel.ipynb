```python
import pandas as pd
import numpy as np

def map_ratetype(ratetype_series):
    
    mapped_values = np.full(len(ratetype_series), 2) 
    
    for i, value in enumerate(ratetype_series):
        if pd.isna(value):
            mapped_values[i] = 2
            continue
            
        if value == "Extra low price! (non-refundable)":
            mapped_values[i] = 0
        elif value == "Extra low price! (non-refundable), Pay at the hotel":
            mapped_values[i] = 1
        elif value == "FREE cancellation":
            mapped_values[i] = 5
        elif "Free cancellation" in value:
            mapped_values[i] = 3
    
    return mapped_values

train['ratetype'] = map_ratetype(train['ratetype'])
test['ratetype'] = map_ratetype(test['ratetype'])

```

```python
test['amenities']=test['amenities'].astype(str).str.len()
train['amenities']=train['amenities'].astype(str).str.len()
test['roomtype']=test['roomtype'].astype(str).str.len()
train['roomtype']=train['roomtype'].astype(str).str.len()
```

```python
train['hotelblock'] = train['hotelblock'].replace('Sold out', 1).fillna(0)
test['hotelblock'] = test['hotelblock'].replace('Sold out', 1).fillna(0)
```

```python
def calculate_meal_score(meal_series):
   
    scores = []
    
    for value in meal_series:
        if pd.isna(value):
            scores.append(0)
            continue
            
        value_str = str(value).lower()
        score = 0
        
        import re
        breakfast_match = re.search(r'breakfast for \{?(\d+)\}?', value_str)
        if breakfast_match:
            score += int(breakfast_match.group(1))
        elif 'breakfast' in value_str:
            score += 1  
            
        if 'lunch' in value_str:
            score += 1
            
        if 'wifi' in value_str or 'wi-fi' in value_str:
            score += 1
            
        if 'parking' in value_str:
            score += 1
            
        scores.append(score)
    
    return scores

train['meal_score'] = calculate_meal_score(train['mealinclusiontype'])
test['meal_score'] = calculate_meal_score(test['mealinclusiontype'])
```

```python
hotel_stats = train.groupby('hotelcode')['rent'].agg(['count', 'mean']).reset_index()
hotel_stats.columns = ['hotelcode', 'hotel_freq', 'hotel_mean_rent']

train = train.merge(hotel_stats[['hotelcode', 'hotel_freq', 'hotel_mean_rent']], on='hotelcode', how='left')
test = test.merge(hotel_stats[['hotelcode', 'hotel_freq', 'hotel_mean_rent']], on='hotelcode', how='left')

test['hotel_freq'] = test['hotel_freq'].fillna(0)
test['hotel_mean_rent'] = test['hotel_mean_rent'].fillna(train['rent'].mean())

train['hotel_freq_log'] = np.log1p(train['hotel_freq'])
test['hotel_freq_log'] = np.log1p(test['hotel_freq'])
```

```python
train=train.drop(columns=['currency','mealinclusiontype','description',"date","hotelcode","id"])
test=test.drop(columns=['currency',"mealinclusiontype",'description',"date","hotelcode","id"])
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.neural_network import MLPRegressor
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import pandas as pd
import numpy as np

def custom_score(y_true, y_pred, num_classes=None):
    """Custom scoring function for regression - using R2 score"""
    return r2_score(y_true=y_true, y_pred=y_pred)

def run_model_with_grid_search(model_name, model, param_grid, x_train, x_test, y_train, y_test, return_model, num_classes=None):
    
    total_combinations = np.prod([len(v) for v in param_grid.values()])
    print(f"Running {model_name} with {total_combinations} model configurations...")
    grid = GridSearchCV(model, param_grid, cv=2, scoring='r2')
    grid.fit(x_train, y_train)
    best_model = grid.best_estimator_
    y_pred = best_model.predict(x_test)
    score = custom_score(y_test, y_pred)
    print(f"{model_name} R2 score: {score:.4f}")
    
    # Additional regression metrics
    mse = mean_squared_error(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    print(f"{model_name} MSE: {mse:.4f}, MAE: {mae:.4f}")
    
    return (score, best_model, y_pred, total_combinations) if return_model else (score, None, None, total_combinations)

def NeuralNetworkBased(x_train, x_test, y_train, y_test, return_model, num_classes=None):
    param_grid = {
        'hidden_layer_sizes': [(50,), (100,), (50, 50), (100, 50)],
        'activation': ['relu', 'tanh'],
        'solver': ['adam'],
        'max_iter': [300],
        'alpha': [0.0001, 0.001]
    }
    return run_model_with_grid_search("NeuralNetworkBased", MLPRegressor(random_state=42), param_grid, x_train, x_test, y_train, y_test, return_model, num_classes)

def SVMBased(x_train, x_test, y_train, y_test, return_model, num_classes=None):
    param_grid = {
        'C': [0.1, 1, 10],
        'kernel': ['linear', 'rbf'],
        'gamma': ['scale', 'auto'],
        'epsilon': [0.1, 0.2]
    }
    return run_model_with_grid_search("SVMBased", SVR(), param_grid, x_train, x_test, y_train, y_test, return_model, num_classes)

def RandomForestBased(x_train, x_test, y_train, y_test, return_model, num_classes=None):
    param_grid = {
        'n_estimators': [50, 100, 200],
        'max_depth': [None, 10, 20],
        'min_samples_split': [2, 5],
        'min_samples_leaf': [1, 2]
    }
    return run_model_with_grid_search("RandomForestBased", RandomForestRegressor(random_state=42), param_grid, x_train, x_test, y_train, y_test, return_model, num_classes)

def XGBoostBased(x_train, x_test, y_train, y_test, return_model, num_classes=None):
    param_grid = {
        'n_estimators': [50, 100, 200],
        'max_depth': [3, 6, 10],
        'learning_rate': [0.01, 0.1, 0.2],
        'subsample': [0.8, 1.0]
    }
    return run_model_with_grid_search("XGBoostBased", XGBRegressor(random_state=42), param_grid, x_train, x_test, y_train, y_test, return_model, num_classes)

def LightGBMBased(x_train, x_test, y_train, y_test, return_model, num_classes=None):
    param_grid = {
        'n_estimators': [100, 200, 300],
        'learning_rate': [0.01, 0.1],
        'num_leaves': [31, 63],
        'max_depth': [-1, 5, 10],
        'subsample': [0.8, 1.0]
    }
    return run_model_with_grid_search("LightGBM", LGBMRegressor(random_state=42, verbose=-1), param_grid, x_train, x_test, y_train, y_test, return_model, num_classes)

def CatBoostBased(x_train, x_test, y_train, y_test, return_model, num_classes=None):
    param_grid = {
        'iterations': [100, 200, 300],
        'learning_rate': [0.01, 0.1],
        'depth': [4, 6, 8],
        'l2_leaf_reg': [1, 3, 5]
    }
    return run_model_with_grid_search("CatBoost", CatBoostRegressor(silent=True, allow_writing_files=False, random_state=42), param_grid, x_train, x_test, y_train, y_test, return_model, num_classes)

def GradientBoostingBased(x_train, x_test, y_train, y_test, return_model, num_classes=None):
    param_grid = {
        'n_estimators': [50, 100, 200],
        'learning_rate': [0.01, 0.1, 0.2],
        'max_depth': [3, 5, 7],
        'subsample': [0.8, 1.0]
    }
    return run_model_with_grid_search("GradientBoosting", GradientBoostingRegressor(random_state=42), param_grid, x_train, x_test, y_train, y_test, return_model, num_classes)

# All these models now work with continuous labels
algorithm_functions_model_training = {
    "RandomForest": RandomForestBased,
    "XGBoost": XGBoostBased,
    "LightGBM": LightGBMBased,
    "CatBoost": CatBoostBased,
    "GradientBoosting": GradientBoostingBased,
    "NeuralNetwork": NeuralNetworkBased,
    "SVM": SVMBased,  # Added back with SVR
}

def model_training(x_train, x_test, y_train, y_test):
    acc_holder = {}
    best_algo = None
    best_accuracy = -float('inf')  # For R2 score, higher is better (can be negative)
    tot_combs_all_models = 0
    
    for name, func in algorithm_functions_model_training.items():
        print(f"\n{'='*50}")
        print(f"Training {name}")
        print(f"{'='*50}")
        
        accuracy, model, y_pred, total_combinations = func(
            x_train.copy(), x_test.copy(), y_train.copy(), y_test.copy(), 
            return_model=False, num_classes=None
        )
        tot_combs_all_models += total_combinations
        acc_holder[name] = accuracy

        if accuracy > best_accuracy:
            best_accuracy = accuracy
            best_algo = name

    # Train the best model with return_model=True
    best_model = None
    best_y_pred = None
    if best_algo:
        print(f"\n{'='*50}")
        print(f"Final training of best model: {best_algo}")
        print(f"{'='*50}")
        accuracy, best_model, best_y_pred, _ = algorithm_functions_model_training[best_algo](
            x_train.copy(), x_test.copy(), y_train.copy(), y_test.copy(), 
            return_model=True, num_classes=None
        )
        
        # Calculate additional metrics for the best model
        mse = mean_squared_error(y_test, best_y_pred)
        mae = mean_absolute_error(y_test, best_y_pred)
        rmse = np.sqrt(mse)
        
        print(f"\nBest Model ({best_algo}) Performance:")
        print(f"R2 Score: {accuracy:.4f}")
        print(f"MSE: {mse:.4f}")
        print(f"RMSE: {rmse:.4f}")
        print(f"MAE: {mae:.4f}")

    return best_model, best_algo, best_accuracy, acc_holder, best_y_pred, tot_combs_all_models

# Additional utility function for regression analysis
def regression_analysis(y_true, y_pred, model_name=""):
    """Comprehensive regression analysis"""
    r2 = r2_score(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    
    # Calculate percentage errors
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    
    print(f"\n{model_name} Regression Analysis:")
    print(f"R² Score: {r2:.4f}")
    print(f"Mean Squared Error: {mse:.4f}")
    print(f"Root Mean Squared Error: {rmse:.4f}")
    print(f"Mean Absolute Error: {mae:.4f}")
    print(f"Mean Absolute Percentage Error: {mape:.2f}%")
    
    return {
        'r2': r2,
        'mse': mse,
        'rmse': rmse,
        'mae': mae,
        'mape': mape
    }
```
